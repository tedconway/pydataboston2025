```python
#emoji2text_off                         # Turn off emoji2text mode temporarily using comment at start of cell while defining dictionary
# Create global emoji-to-text Python dictionary, set global indicator that translation is active (only needs to run once)
global emoji2text_dict, emoji2txt
emoji2text_dict = {
'🤔':'if', '❎':'else', '🖨️':'print',
'🐼':'pandas', '🦆':'duckdb', '📈':'plotly', '𝄜':'excel',
'🔤':'str', '💾':'data', '📅':'date', '🕙':'time', '🔄':'while',
'🛢':'create table', '🗑️':'drop table', '🛒':'select', '⬅️':'from',
'🔗':'join', '↕️':'order by', '⬇️':'asc', '⬇️':'desc', '∑':'group by',
'🚗':'cars'}
globals()['emoji2txt']=True             # With dictionary defined, turn emoji2text mode on via global variable emoji2txt
```

```python
# Define/activate IPyton Input Transformer to translate emojis in submitted cells into text for Python (only needs to run once)
import regex, json, pandas as pd

# Parse submitted Python code for emojis, other characters
def grapheme_positions(text):
    matches = regex.finditer(r'\X', text)
    return [{'grapheme': m.group()} for m in matches]

# Process submitted Python code, replacing emojis with assoicated code from global dictionary
def emoji2text(lines):
    global emoji2txt
    global emoji2text_dict
    emoji2text_off=False
    if "".join(lines[0][:15])=="#emoji2text_off": emoji2text_off=True
    if emoji2text_off==True or 'emoji2txt' not in globals():
        return lines
    if emoji2txt==False:
        return lines
    df = pd.DataFrame.from_dict(grapheme_positions("".join(lines)), orient='columns')
    df['replstr'] = df['grapheme'].map(emoji2text_dict)
    df['replstr'] = df['replstr'].fillna(df['grapheme'])
    return "".join(df["replstr"].tolist()).splitlines(keepends=True)

# Add emoji2text input transformer after deleting any existing instances
ip = get_ipython();
for i in range(len(ip.input_transformers_cleanup)-1,-1,-1):
  if type(ip.input_transformers_cleanup[i]).__name__=='function':
    if ip.input_transformers_cleanup[i].__name__=='emoji2text':
      del ip.input_transformers_cleanup[i]
ip = get_ipython()
ip.input_transformers_cleanup.append(emoji2text)
```
